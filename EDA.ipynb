```python
#Scoring Model and Rating Scoe
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. Load dataset
df = pd.read_csv("company_full_metrics.csv")

# -----------------------------
# 2. Define metric groups
financial_metrics = [
    "Stock_Price", "Market_Cap", "Revenue_Growth", 
    "Earnings_Per_Share", "PE_Ratio", 
    "Debt_to_Equity Ratio", "Dividend_Yield"
]

esg_metrics = [
    "ESG_SCORE", "Environmental Risk Score", 
    "Renewable_Energy_pct", "Governance_Score", 
    "Social Risk Score"
]

# -----------------------------
# 3. Normalize metrics (0–1 scaling)
scaler = MinMaxScaler()

df_fin_scaled = pd.DataFrame(
    scaler.fit_transform(df[financial_metrics]),
    columns=financial_metrics
)

df_esg_scaled = pd.DataFrame(
    scaler.fit_transform(df[esg_metrics]),
    columns=esg_metrics
)

# -----------------------------
# 4. Invert “risk” metrics (lower = better)
invert_cols = ["Environmental Risk Score", "Social Risk Score"]
for col in invert_cols:
    df_esg_scaled[col] = 1 - df_esg_scaled[col]

# -----------------------------
# 5. Compute Sub-Scores
df["Finance_Score"] = df_fin_scaled.mean(axis=1)
df["ESG_Score"] = df_esg_scaled.mean(axis=1)

# -----------------------------
# 6. Final Composite Scores
df["FinalScore_50_50"] = 0.5 * df["Finance_Score"] + 0.5 * df["ESG_Score"]
df["FinalScore_60_40"] = 0.6 * df["Finance_Score"] + 0.4 * df["ESG_Score"]

# -----------------------------
# 7. Ranking
df["Rank_50_50"] = df["FinalScore_50_50"].rank(ascending=False)
df["Rank_60_40"] = df["FinalScore_60_40"].rank(ascending=False)

# -----------------------------
# 8. Output Ranked Data
ranked_df = df.sort_values("FinalScore_50_50", ascending=False)

# Save to CSV
ranked_df.to_csv("company_ranked_scores.csv", index=False)

print(ranked_df[["Company", "FinalScore_50_50", "Rank_50_50", "FinalScore_60_40", "Rank_60_40"]])
```

```python
import pandas as pd

# Load your cleaned dataset
df = pd.read_csv("company_full_metrics.csv")

# --- Recompute normalized metrics and scores (same as earlier step) ---
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

# Finance metrics
finance_metrics = ["Stock_Price","Market_Cap","Revenue_Growth","Earnings_Per_Share","PE_Ratio","Dividend_Yield"]
df_finance_scaled = pd.DataFrame(scaler.fit_transform(df[finance_metrics]), columns=finance_metrics)

# ESG metrics
esg_metrics = ["ESG_SCORE","Environmental Risk Score","Renewable_Energy_pct","Governance_Score","Social Risk Score"]
df_esg_scaled = pd.DataFrame(scaler.fit_transform(df[esg_metrics]), columns=esg_metrics)

# Add back company column
df_finance_scaled["Company"] = df["Company"]
df_esg_scaled["Company"] = df["Company"]

# Compute composite scores
df["Finance_Score"] = df_finance_scaled[finance_metrics].mean(axis=1)
df["ESG_Score"] = df_esg_scaled[esg_metrics].mean(axis=1)
df["FinalScore_50_50"] = 0.5 * df["Finance_Score"] + 0.5 * df["ESG_Score"]
df["FinalScore_60_40"] = 0.6 * df["Finance_Score"] + 0.4 * df["ESG_Score"]

# Ranking
df["Rank_50_50"] = df["FinalScore_50_50"].rank(ascending=False)
df["Rank_60_40"] = df["FinalScore_60_40"].rank(ascending=False)

# This is your main dataframe for visualization
df_scores = df
```

```python
#A — Summary table: top / bottom groups + means
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("company_full_metrics.csv")      # your cleaned file
scores = pd.read_csv("company_ranked_scores.csv") # file saved earlier if any
df = df.merge(scores[['Company','FinalScore_50_50','FinalScore_60_40','Finance_Score','ESG_Score']], on='Company')

# Top 5 and bottom 5 by FinalScore_50_50
top5 = df.sort_values('FinalScore_50_50', ascending=False).head(5)
bot5 = df.sort_values('FinalScore_50_50', ascending=False).tail(5)

print("Top 5\n", top5[['Company','FinalScore_50_50','Finance_Score','ESG_Score']])
print("\nBottom 5\n", bot5[['Company','FinalScore_50_50','Finance_Score','ESG_Score']])

# Means for top vs bottom groups
metrics = ['Revenue_Growth','ESG_SCORE','Renewable_Energy_pct','PE_Ratio','Dividend_Yield','Market_Cap']
print("\nMean comparison (top5 vs bottom5):")
print("Top5 means:\n", top5[metrics].apply(np.mean))
print("Bottom5 means:\n", bot5[metrics].apply(np.mean))
```

```python
#Significance test (to check if top and bottom groups statistically different?)
# Example: test ESG_SCORE difference
top_vals = top5['ESG_SCORE']
bot_vals = bot5['ESG_SCORE']

# normality check
print(stats.shapiro(top_vals))
print(stats.shapiro(bot_vals))

# if normal:
tstat, pval = stats.ttest_ind(top_vals, bot_vals, equal_var=False)
print("t-test ESG_SCORE top vs bottom:", tstat, pval)

# if not normal:
u_stat, p_u = stats.mannwhitneyu(top_vals, bot_vals, alternative='two-sided')
print("Mann-Whitney ESG_SCORE top vs bottom:", u_stat, p_u)
```

```python
#Sensitivity analysis (varying weights)
weights = [(0.5,0.5),(0.6,0.4),(0.4,0.6),(0.7,0.3)]
results = []
for w_fin, w_esg in weights:
    df['Final'] = w_fin * df['Finance_Score'] + w_esg * df['ESG_Score']
    df['Rank'] = df['Final'].rank(ascending=False)
    top = df.sort_values('Final', ascending=False).head(5)['Company'].tolist()
    results.append({'weights':f'{w_fin}-{w_esg}','top5':top})
pd.DataFrame(results)
```

```python
#Automating Scoring + Ranking
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# ---------------------------
# 1. Load Data
# ---------------------------
df = pd.read_csv("company_full_metrics.csv")   # replace with your file name

# Standardize column names (fix spaces & consistency)
df.rename(columns={
    "Debt_to_Equity Ratio": "Debt_to_Equity_Ratio",
    "Environmental Risk Score": "Environmental_Risk_Score",
    "Social Risk Score": "Social_Risk_Score"
}, inplace=True)

# ---------------------------
# 2. Define Metrics
# ---------------------------
metrics = [
    "ESG_SCORE",
    "Environmental_Risk_Score",
    "Renewable_Energy_pct",
    "Governance_Score",
    "Stock_Price",
    "Market_Cap",
    "Revenue_Growth",
    "Earnings_Per_Share",
    "PE_Ratio",
    "Debt_to_Equity_Ratio",
    "Social_Risk_Score",
    "Dividend_Yield"
]

esg_metrics = [
    "ESG_SCORE",
    "Environmental_Risk_Score",
    "Renewable_Energy_pct",
    "Governance_Score",
    "Social_Risk_Score"
]

financial_metrics = [
    "Stock_Price",
    "Market_Cap",
    "Revenue_Growth",
    "Earnings_Per_Share",
    "PE_Ratio",
    "Debt_to_Equity_Ratio",
    "Dividend_Yield"
]

# ---------------------------
# 3. Normalization
# ---------------------------
scaler = MinMaxScaler()
df_scaled = df.copy()
df_scaled[metrics] = scaler.fit_transform(df[metrics])

# ---------------------------
# 4. Composite Scoring
# ---------------------------
df_scaled["ESG_Score_Avg"] = df_scaled[esg_metrics].mean(axis=1)
df_scaled["Financial_Score"] = df_scaled[financial_metrics].mean(axis=1)

df_scaled["FinalScore_50_50"] = (
    0.5 * df_scaled["ESG_Score_Avg"] + 0.5 * df_scaled["Financial_Score"]
)
df_scaled["FinalScore_60_40"] = (
    0.6 * df_scaled["ESG_Score_Avg"] + 0.4 * df_scaled["Financial_Score"]
)

# ---------------------------
# 5. Ranking
# ---------------------------
df_scaled["Rank_50_50"] = df_scaled["FinalScore_50_50"].rank(ascending=False).astype(int)
df_scaled["Rank_60_40"] = df_scaled["FinalScore_60_40"].rank(ascending=False).astype(int)

# ---------------------------
# 6. Save Results
# ---------------------------
df_scaled.to_csv("normalized_scores_companies.csv", index=False)

print("✅ Scoring + Ranking complete. Results saved to 'normalized_scores_companies.csv'")
```
